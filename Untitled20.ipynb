```python
#q1

import sqlite3

def create_database():
    try:
        with sqlite3.connect('university.db') as conn:
            cursor = conn.cursor()

            cursor.execute('''
                CREATE TABLE IF NOT EXISTS students (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    first_name TEXT NOT NULL,
                    last_name TEXT NOT NULL,
                    group_name TEXT NOT NULL,
                    admission_year INTEGER NOT NULL,
                    average_grade REAL,
                    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
                )
            ''')

            cursor.execute('''
                CREATE TABLE IF NOT EXISTS courses (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    course_name TEXT UNIQUE NOT NULL,
                    instructor TEXT NOT NULL,
                    credits INTEGER NOT NULL,
                    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
                )
            ''')

            print("База данных и таблицы успешно созданы!")

    except sqlite3.Error as e:
        print(f"Ошибка при создании базы данных: {e}")

if __name__ == "__main__":
    create_database()
```

```python
#Q3

import sqlite3
from typing import List, Dict, Any, Optional

def add_student(first_name: str, last_name: str, group_name: str,
                admission_year: int, average_grade: Optional[float] = None) -> Optional[int]:
    try:
        with sqlite3.connect('university.db') as conn:
            cursor = conn.cursor()

            cursor.execute('''
                INSERT INTO students (first_name, last_name, group_name, admission_year, average_grade)
                VALUES (?, ?, ?, ?, ?)
            ''', (first_name, last_name, group_name, admission_year, average_grade))

            conn.commit()
            return cursor.lastrowid

    except sqlite3.Error as e:
        print(f"Error adding student: {e}")
        return None

# table student_courses
def create_student_courses_table() -> None:

    try:
        with sqlite3.connect('university.db') as conn:
            cursor = conn.cursor()

            cursor.execute('''
                CREATE TABLE IF NOT EXISTS student_courses (
                    student_id INTEGER NOT NULL,
                    course_id INTEGER NOT NULL,
                    enrolled_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
                    PRIMARY KEY (student_id, course_id),
                    FOREIGN KEY (student_id) REFERENCES students(id) ON DELETE CASCADE,
                    FOREIGN KEY (course_id) REFERENCES courses(id) ON DELETE CASCADE
                )
            ''')

    except sqlite3.Error as e:
        print(f"Error creating table: {e}")

#courses
def add_course(course_name: str, instructor: str, credits: int) -> Optional[int]:
    try:
        with sqlite3.connect('university.db') as conn:
            cursor = conn.cursor()

            cursor.execute('''
                INSERT INTO courses (course_name, instructor, credits)
                VALUES (?, ?, ?)
            ''', (course_name, instructor, credits))

            conn.commit()
            return cursor.lastrowid

    except sqlite3.Error as e:
        print(f"Error adding course: {e}")
        return None

def enroll_student_in_course(student_id: int, course_id: int) -> bool:

    try:
        with sqlite3.connect('university.db') as conn:
            cursor = conn.cursor()

            # существование студента и курса
            cursor.execute('SELECT id FROM students WHERE id = ?', (student_id,))
            if not cursor.fetchone():
                print(f"Student {student_id} does not exist")
                return False

            cursor.execute('SELECT id FROM courses WHERE id = ?', (course_id,))
            if not cursor.fetchone():
                print(f"Course {course_id} does not exist")
                return False

            #запись на курсы
            cursor.execute('''
                INSERT OR IGNORE INTO student_courses (student_id, course_id)
                VALUES (?, ?)
            ''', (student_id, course_id))

            conn.commit()
            return True

    except sqlite3.Error as e:
        print(f"Error enrolling student: {e}")
        return False

def get_student_courses(student_id: int) -> List[Dict[str, Any]]:
    try:
        with sqlite3.connect('university.db') as conn:
            conn.row_factory = sqlite3.Row
            cursor = conn.cursor()

            cursor.execute('''
                SELECT c.* FROM courses c
                JOIN student_courses sc ON c.id = sc.course_id
                WHERE sc.student_id = ?
            ''', (student_id,))

            return [dict(row) for row in cursor.fetchall()]

    except sqlite3.Error as e:
        print(f"Error getting student courses: {e}")
        return []

def get_course_students(course_id: int) -> List[Dict[str, Any]]:
    try:
        with sqlite3.connect('university.db') as conn:
            conn.row_factory = sqlite3.Row
            cursor = conn.cursor()

            cursor.execute('''
                SELECT s.* FROM students s
                JOIN student_courses sc ON s.id = sc.student_id
                WHERE sc.course_id = ?
            ''', (course_id,))

            return [dict(row) for row in cursor.fetchall()]

    except sqlite3.Error as e:
        print(f"Error getting course students: {e}")
        return []

def transfer_student(student_id: int, new_group: str) -> bool:

    try:
        with sqlite3.connect('university.db') as conn:
            # транзакция
            cursor = conn.cursor()

            #проверка студента
            cursor.execute('SELECT group_name FROM students WHERE id = ?', (student_id,))
            result = cursor.fetchone()
            if not result:
                print(f"Student {student_id} does not exist")
                return False

            old_group = result[0]

            #если группы совпадают
            if old_group == new_group:
                print(f"Student already in group {new_group}")
                return False

            #обновление группы
            cursor.execute('UPDATE students SET group_name = ? WHERE id = ?',
                         (new_group, student_id))

            #удаление всех курсов студента
            cursor.execute('DELETE FROM student_courses WHERE student_id = ?', (student_id,))

            conn.commit()
            print(f"Student {student_id} moved from {old_group} to {new_group}")
            return True

    except sqlite3.Error as e:
        print(f"Error transferring student: {e}")
        return False

#testing
def test_all_functions():

    create_student_courses_table()


    # очиста таблиц
    try:
        with sqlite3.connect('university.db') as conn:
            cursor = conn.cursor()
            cursor.execute('DELETE FROM students')
            cursor.execute('DELETE FROM courses')
            cursor.execute('DELETE FROM student_courses')
            conn.commit()
    except:
        pass

    print("\nAdding students")
    student1_id = add_student("Aziz", "Ashwariya", "SS-99", 2022, 4.5)
    student2_id = add_student("Ahmad", "Ahmed", "GR-01", 2022, 4.8)
    student3_id = add_student("Yasmin", "Raj", "GR-02", 2024, 3.9)

    print(f"   Students added: Aziz (ID {student1_id}), Ahmad (ID {student2_id}), Yasmin (ID {student3_id})")

    print("\nAdding courses")
    course1_id = add_course("Arabic", "Prof. Smith", 5)
    course2_id = add_course("Hindi", "Prof. Johnson", 4)
    course3_id = add_course("Math", "Prof. Williams", 6)

    print(f"   Courses added: IDs {course1_id}, {course2_id}, {course3_id}")

    print("\nEnrolling students")

    # Aziz (ID 1) на курс 1 и 3
    if enroll_student_in_course(student1_id, course1_id):
        print("   Aziz enrolled in Arabic")
    if enroll_student_in_course(student1_id, course3_id):
        print("   Aziz enrolled in Math")

    # Yasmin (ID 3) на курс 2
    if enroll_student_in_course(student3_id, course2_id):
        print("   Yasmin enrolled in Hindi")

    # azizs courses
    print("\n Aziz's courses:")
    aziz_courses = get_student_courses(student1_id)
    if aziz_courses:
        for course in aziz_courses:
            print(f"   - {course['course_name']} ({course['credits']} credits)")
    else:
        print("   No courses found")

    print("\nStudents in Arabic course:")
    prog_students = get_course_students(course3_id)
    if prog_students:
        for student in prog_students:
            print(f"   - {student['first_name']} {student['last_name']} (Group: {student['group_name']})")
    else:
        print("   No students found")

    # transferring
    print("\nTransferring Aziz to GR-03")
    if transfer_student(student1_id, "GR-03"):
        print("   Transfer successful")

    # курсы после перевода
    print("\n Aziz's courses after transfer:")
    aziz_new_courses = get_student_courses(student1_id)
    if aziz_new_courses:
        for course in aziz_new_courses:
            print(f"   - {course['course_name']}")
    else:
        print("   No courses (removed during transfer)")

    # Azizs group
    print("\nChecking Aziz's new group:")
    try:
        with sqlite3.connect('university.db') as conn:
            cursor = conn.cursor()
            cursor.execute('SELECT group_name FROM students WHERE id = ?', (student1_id,))
            result = cursor.fetchone()
            if result:
                print(f"   Aziz is now in group: {result[0]}")
    except:
        pass

    print("\nAll tests completed!")

#start
if __name__ == "__main__":
    test_all_functions()
```

```python
#Q4

import sqlite3
from typing import Optional, List, Dict, Any, Tuple

class UniversityDB:
    "Класс-обертка для работы с базой данных университета"

    def __init__(self, db_path: str = 'university.db'):
        self.db_path = db_path
        self.connection = None
        self.cursor = None

    def __enter__(self):
        self.connection = sqlite3.connect(self.db_path)
        self.connection.row_factory = sqlite3.Row
        self.cursor = self.connection.cursor()
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        if self.connection:
            if exc_type is None:
                self.connection.commit()
            else:
                self.connection.rollback()
            self.connection.close()

    def execute_query(self, query: str, params: Optional[Tuple] = None) -> int:
        try:
            if params:
                self.cursor.execute(query, params)
            else:
                self.cursor.execute(query)
            return self.cursor.rowcount
        except sqlite3.Error as e:
            print(f"Error: {e}")
            return 0

    def fetch_all(self, query: str, params: Optional[Tuple] = None) -> List[Dict[str, Any]]:
        try:
            if params:
                self.cursor.execute(query, params)
            else:
                self.cursor.execute(query)
            rows = self.cursor.fetchall()
            return [dict(row) for row in rows]
        except sqlite3.Error as e:
            print(f"Error: {e}")
            return []

    def fetch_one(self, query: str, params: Optional[Tuple] = None) -> Optional[Dict[str, Any]]:
        try:
            if params:
                self.cursor.execute(query, params)
            else:
                self.cursor.execute(query)
            row = self.cursor.fetchone()
            return dict(row) if row else None
        except sqlite3.Error as e:
            print(f"Error: {e}")
            return None


    def get_student_statistics(self) -> Dict[str, Any]:

        try:
            # всего студентов
            total_result = self.fetch_one("SELECT COUNT(*) as total FROM students")
            total_students = total_result['total'] if total_result else 0

            # Ср балл
            avg_result = self.fetch_one("SELECT AVG(average_grade) as avg FROM students WHERE average_grade IS NOT NULL")
            avg_grade = round(avg_result['avg'], 2) if avg_result and avg_result['avg'] else 0.0

            # статистика по группам
            group_stats = self.fetch_all("""
                SELECT group_name, COUNT(*) as count,
                       AVG(average_grade) as avg_grade
                FROM students
                GROUP BY group_name
                ORDER BY group_name
            """)

            return {
                'total_students': total_students,
                'avg_grade': avg_grade,
                'group_stats': group_stats
            }

        except Exception as e:
            print(f"Error: {e}")
            return {'total_students': 0, 'avg_grade': 0.0, 'group_stats': []}

    def get_top_students(self, limit: int = 5) -> List[Dict[str, Any]]:

        try:
            query = """
                SELECT id, first_name, last_name, group_name, average_grade
                FROM students
                WHERE average_grade IS NOT NULL
                ORDER BY average_grade DESC
                LIMIT ?
            """
            return self.fetch_all(query, (limit,))
        except Exception as e:
            print(f"Error: {e}")
            return []

    def get_group_statistics(self) -> List[Dict[str, Any]]:

        try:
            result = self.fetch_all("""
                SELECT
                    group_name,
                    COUNT(*) as student_count,
                    AVG(average_grade) as avg_grade,
                    MIN(admission_year) as oldest_year,
                    MAX(admission_year) as newest_year
                FROM students
                GROUP BY group_name
                ORDER BY group_name
            """)

            # форматирование числа
            for row in result:
                if row['avg_grade']:
                    row['avg_grade'] = round(row['avg_grade'], 2)

            return result
        except Exception as e:
            print(f"Error: {e}")
            return []


def main():
    print("UniversityDB Demo")
    print("=" * 40)

    with UniversityDB('university.db') as db:

        test = db.fetch_one("SELECT COUNT(*) as count FROM students")
        print(f"   Students in DB: {test['count'] if test else 0}")

        print("\nStudent statistics:")
        stats = db.get_student_statistics()
        print(f"   Total students: {stats['total_students']}")
        print(f"   Average grade: {stats['avg_grade']}")

        print("\nTop 3 students:")
        top = db.get_top_students(3)
        for i, student in enumerate(top, 1):
            grade = student['average_grade'] if student['average_grade'] else "N/A"
            print(f"   {i}. {student['first_name']} {student['last_name']} - {grade}")

        print("\nGroup statistics:")
        groups = db.get_group_statistics()
        for group in groups:
            print(f"   Group {group['group_name']}:")
            print(f"     Students: {group['student_count']}")
            print(f"     Avg grade: {group['avg_grade']}")
            print(f"     Years: {group['oldest_year']}-{group['newest_year']}")

        # пример базовых методов
        print("\nBasic methods example:")

        # fetch_one
        student = db.fetch_one("SELECT * FROM students WHERE id = ?", (10,))
        if student:
            print(f"   Student ID 10: {student['first_name']} {student['last_name']}")

        # fetch_all
        students = db.fetch_all("SELECT first_name, last_name FROM students LIMIT 2")
        print(f"   First 2 students: {len(students)} found")

        # execute_query
        count = db.execute_query("UPDATE students SET group_name = ? WHERE id = ?", ("TEST", 10))
        print(f"   Updated {count} rows")

if __name__ == "__main__":
    main()
```

```python
#Q5

from typing import Optional
import sqlite3

class UniversityDB:
    def __init__(self, db_path='university.db'):
        self.db_path = db_path

    def __enter__(self):
        self.conn = sqlite3.connect(self.db_path)
        self.conn.row_factory = sqlite3.Row
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        if self.conn:
            if exc_type is None:
                self.conn.commit()
            else:
                self.conn.rollback()
            self.conn.close()

    def execute(self, query, params=None):
        cursor = self.conn.cursor()
        try:
            if params:
                cursor.execute(query, params)
            else:
                cursor.execute(query)
            return cursor
        except sqlite3.Error as e:
            print(f"Error: {e}")
            return None

    def add_student(self, first_name, last_name, group_name, admission_year, grade=None):
        query = """
            INSERT INTO students (first_name, last_name, group_name, admission_year, average_grade)
            VALUES (?, ?, ?, ?, ?)
        """
        cursor = self.execute(query, (first_name, last_name, group_name, admission_year, grade))
        return cursor.lastrowid if cursor else None

    def get_all_students(self):
        cursor = self.execute("SELECT * FROM students ORDER BY last_name, first_name")
        return [dict(row) for row in cursor.fetchall()] if cursor else []

    def search_students(self, last_name_part):
        query = "SELECT * FROM students WHERE last_name LIKE ? ORDER BY last_name, first_name"
        cursor = self.execute(query, (f"%{last_name_part}%",))
        return [dict(row) for row in cursor.fetchall()] if cursor else []

    def update_grade(self, student_id, new_grade):
        query = "UPDATE students SET average_grade = ? WHERE id = ?"
        cursor = self.execute(query, (new_grade, student_id))
        return cursor.rowcount > 0 if cursor else False

    def delete_student(self, student_id):
        query = "DELETE FROM students WHERE id = ?"
        cursor = self.execute(query, (student_id,))
        return cursor.rowcount > 0 if cursor else False

    def get_statistics(self):
        cursor = self.execute("SELECT COUNT(*) as total FROM students")
        total = cursor.fetchone()[0] if cursor else 0

        cursor = self.execute("SELECT AVG(average_grade) as avg FROM students WHERE average_grade IS NOT NULL")
        avg_row = cursor.fetchone() if cursor else None
        avg_grade = round(avg_row[0], 2) if avg_row and avg_row[0] else 0.0

        return {'total': total, 'avg_grade': avg_grade}

    def get_top_students(self, limit=5):
        query = """
            SELECT id, first_name, last_name, group_name, average_grade
            FROM students
            WHERE average_grade IS NOT NULL
            ORDER BY average_grade DESC
            LIMIT ?
        """
        cursor = self.execute(query, (limit,))
        return [dict(row) for row in cursor.fetchall()] if cursor else []

# Функции интерфейса
def show_students(students):
    if not students:
        print("Нет студентов")
        return

    print(f"\n{'ID':<4} {'Имя':<15} {'Фамилия':<15} {'Группа':<10} {'Год':<6} {'Оценка':<8}")
    print("-" * 60)

    for s in students:
        grade = s.get('average_grade', '')
        grade_str = f"{grade:.2f}" if grade else "Н/Д"
        print(f"{s['id']:<4} {s['first_name']:<15} {s['last_name']:<15} "
              f"{s['group_name']:<10} {s['admission_year']:<6} {grade_str:<8}")

    print(f"Всего: {len(students)} студентов")

def add_student():
    print("\nДобавление студента")

    first_name = input("Имя: ").strip()
    while not first_name:
        print("Имя обязательно")
        first_name = input("Имя: ").strip()

    last_name = input("Фамилия: ").strip()
    while not last_name:
        print("Фамилия обязательна")
        last_name = input("Фамилия: ").strip()

    group_name = input("Группа: ").strip()
    while not group_name:
        print("Группа обязательна")
        group_name = input("Группа: ").strip()

    # Валидация года
    while True:
        try:
            year = int(input("Год поступления: "))
            if 2000 <= year <= 2030:
                break
            print("Год должен быть 2000-2030")
        except:
            print("Введите число")

    # Валидация оценки
    grade = None
    while True:
        grade_input = input("Оценка (Enter если нет): ").strip()
        if not grade_input:
            break
        try:
            grade = float(grade_input)
            if 0 <= grade <= 5:
                break
            print("Оценка должна быть 0-5")
        except:
            print("Введите число")

    with UniversityDB() as db:
        student_id = db.add_student(first_name, last_name, group_name, year, grade)
        if student_id:
            print(f"Студент добавлен. ID: {student_id}")
        else:
            print("Ошибка")

def view_all_students():
    print("\n Все студенты")
    with UniversityDB() as db:
        students = db.get_all_students()
        show_students(students)

def search_student():
    print("\n Поиск студента")
    name = input("Фамилия (часть): ").strip()
    if not name:
        print("Введите фамилию")
        return

    with UniversityDB() as db:
        students = db.search_students(name)
        if students:
            print(f"Найдено {len(students)} студентов:")
            show_students(students)
        else:
            print("Студенты не найдены")

def update_grade():
    print("\nОбновление оценки")

    try:
        student_id = int(input("ID студента: "))
    except:
        print("ID должен быть числом")
        return

    with UniversityDB() as db:
        students = db.search_students("")
        ids = [s['id'] for s in students]
        if student_id not in ids:
            print(f"Студент с ID {student_id} не найден")
            return

        while True:
            try:
                new_grade = float(input("Новая оценка (0-5): "))
                if 0 <= new_grade <= 5:
                    break
                print("Оценка должна быть 0-5")
            except:
                print("Введите число")

        if db.update_grade(student_id, new_grade):
            print(" Оценка обновлена")
        else:
            print("Ошибка")

def delete_student():
    print("\nУдаление студента ")

    try:
        student_id = int(input("ID студента для удаления: "))
    except:
        print("ID должен быть числом")
        return

    confirm = input(f"Удалить студента {student_id}? (да/нет): ").lower()
    if confirm == 'да':
        with UniversityDB() as db:
            if db.delete_student(student_id):
                print("Студент удален")
            else:
                print("Ошибка или студент не найден")
    else:
        print("Отменено")

def show_stats():
    print("\n Статистика")

    with UniversityDB() as db:
        stats = db.get_statistics()
        top = db.get_top_students(5)

        print(f"Всего студентов: {stats['total']}")
        print(f"Средний балл: {stats['avg_grade']}")

        if top:
            print("\nТоп 5 студентов:")
            for i, s in enumerate(top, 1):
                print(f"{i}. {s['first_name']} {s['last_name']} - {s['average_grade']}")
        else:
            print("\nНет студентов с оценками")

# Главное меню
def main():
    # Создаем таблицу если нет
    with UniversityDB() as db:
        db.execute("""
            CREATE TABLE IF NOT EXISTS students (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                first_name TEXT NOT NULL,
                last_name TEXT NOT NULL,
                group_name TEXT NOT NULL,
                admission_year INTEGER NOT NULL,
                average_grade REAL,
                created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
            )
        """)

    while True:
        print("\n" + "=" * 40)
        print(" УНИВЕРСИТЕТСКИЙ УЧЕТ")
        print("=" * 40)
        print("1. Добавить студента")
        print("2. Все студенты")
        print("3. Найти студента")
        print("4. Обновить оценку")
        print("5. Удалить студента")
        print("6. Статистика")
        print("7. Выход")
        print("=" * 40)

        choice = input("Выберите: ")

        if choice == '1':
            add_student()
        elif choice == '2':
            view_all_students()
        elif choice == '3':
            search_student()
        elif choice == '4':
            update_grade()
        elif choice == '5':
            delete_student()
        elif choice == '6':
            show_stats()
        elif choice == '7':
            print("Выход")
            break
        else:
            print("Неверный выбор")

if __name__ == "__main__":
    main()
```
