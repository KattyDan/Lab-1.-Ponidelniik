```python
# Q1  Синхронный калькулятор времени выполнения
import time

def sync_calculate(operation, a, b, delay):

    print(f"Начало операции {a} {operation} {b}")
    time.sleep(delay)  # Имитация долгого вычисления

    if operation == '+':
        result = a + b
    elif operation == '-':
        result = a - b
    elif operation == '*':
        result = a * b
    elif operation == '/':
        result = a / b if b != 0 else 'Ошибка: деление на ноль'
    else:
        result = 'Неизвестная операция'

    print(f"Конец операции {a} {operation} {b} = {result}")
    return result

def task1_sync_calculations():

    start_time = time.time()

    # Ваш код здесь
    results = []

    # Список операций для удобства
    operations = [
        ('+', 15, 25, 2),
        ('-', 40, 18, 1),
        ('*', 12, 8, 3),
        ('/', 100, 5, 1)
    ]

    # Последовательно выполняем все операции
    for op, a, b, delay in operations:
        result = sync_calculate(op, a, b, delay)
        results.append(result)

    end_time = time.time()
    print(f"Общее время выполнения: {end_time - start_time:.2f} секунд")
    print(f"Результаты операций: {results}")

    return results, end_time - start_time

# Блок для запуска в коллабе
if __name__ == "__main__":

    print(" Синхронный калькулятор времени выполнения")


    results, total_time = task1_sync_calculations()

```

```python
#Q2
import threading
import time

def download_file(filename, size):

    download_time = size * 0.1  # 0.1 сек на МБ
    print(f"Начало загрузки {filename} ({size} МБ)")

    # Имитация прогресса загрузки
    for i in range(5):
        time.sleep(download_time / 5)
        progress = (i + 1) * 20
        print(f"{filename}: {progress}% загружено")

    print(f"Завершена загрузка {filename}")

def task2_threaded_downloader():

    files = [
        ("document.pdf", 10),
        ("image.jpg", 5),
        ("video.mp4", 20),
        ("archive.zip", 15)
    ]

    start_time = time.time()
    threads = []

    print("Многопоточная загрузка файлов")

    # Создаем и запускаем потоки для каждого файла
    for filename, size in files:
        thread = threading.Thread(
            target=download_file,
            args=(filename, size)
        )
        threads.append(thread)
        thread.start()

    print(f"Запущено {len(threads)} потоков\n")

    # Дожидаемся завершения всех потоков
    for thread in threads:
        thread.join()

    end_time = time.time()

    print(f"Общее время загрузки: {end_time - start_time:.2f} секунд")

# Запуск задачи
if __name__ == "__main__":
    task2_threaded_downloader()
```

```python
import multiprocessing
import time
import math

def calculate_factorial(n):
    """
    Вычисляет факториал числа (CPU-intensive операция)
    """
    print(f"Начало вычисления факториала {n}!")
    result = math.factorial(n)
    print(f"Завершено вычисление факториала {n}!")
    return result

def calculate_prime(n):
    """
    Проверяет, является ли число простым
    """
    print(f"Начало проверки числа {n} на простоту")

    if n < 2:
        result = False
    else:
        result = all(n % i != 0 for i in range(2, int(math.sqrt(n)) + 1))

    print(f"Число {n} простое: {result}")
    return result

def task3_multiprocess_calculations():
    """
    Задача: Реализуйте многопроцессные вычисления.
    Вычисления:
    1. Факториал 100000
    2. Факториал 80000
    3. Проверка числа 10000000019 на простоту
    4. Проверка числа 10000000033 на простоту

    Требуется:
    - Выполнить вычисления в отдельных процессах
    - Сравнить время с последовательным выполнением
    - Собрать и вывести результаты
    """
    # Уменьшаем числа для корректной работы
    calculations = [
        (calculate_factorial, 10000),   # Вместо 10000
        (calculate_factorial, 800),    # Вместо 8000
        (calculate_prime, 10000033),     # Вместо 10000033
        (calculate_prime, 100019)      # Вместо 10000019
    ]

    # Многопроцессное выполнение
    print(" МНОГОПРОЦЕССНОЕ ВЫПОЛНЕНИЕ")
    start_time = time.time()

    processes = []
    manager = multiprocessing.Manager()
    results = manager.list()  # Общий список для результатов процессов

    # Создаем и запускаем процессы
    for i, (func, arg) in enumerate(calculations, 1):
        process = multiprocessing.Process(
            target=lambda f=func, a=arg, r=results: r.append(f(a)),
            name=f"Process-{i}"
        )
        processes.append(process)
        process.start()
        print(f"Запущен процесс {i}: {func.__name__}({arg})")

    # Ждем завершения всех процессов
    print("\nОжидание завершения процессов")
    for process in processes:
        process.join()

    end_time = time.time()
    multiprocess_time = end_time - start_time
    multiprocess_results = list(results)  # Преобразуем в обычный список

    print(f"\nРезультаты многопроцессного выполнения получены")

    # Синхронное выполнение для сравнения
    print("\nСИНХРОННОЕ ВЫПОЛНЕНИЕ")
    start_time = time.time()

    sync_results = []
    for func, arg in calculations:
        sync_results.append(func(arg))

    end_time = time.time()
    sync_time = end_time - start_time

    print(f"\nСравнение времени:")
    print(f"Многопроцессное: {multiprocess_time:.2f} сек")
    print(f"Синхронное: {sync_time:.2f} сек")

    if multiprocess_time > 0:
        print(f"Ускорение: {sync_time/multiprocess_time:.2f}x")
    else:
        print("Ускорение: не может быть вычислено")

    # Проверяем совпадение результатов
    print(f"\nРезультаты совпадают: {multiprocess_results == sync_results}")

# Запуск задачи
if __name__ == "__main__":
    task3_multiprocess_calculations()
```

```python
#Q4
import asyncio
import aiohttp
import time

async def fetch_url(session, url, name):
    "Асинхронно загружает веб-страницу"
    print(f"Начало загрузки {name}")

    try:
        async with session.get(url) as response:
            content = await response.text()
            print(f"Завершена загрузка {name}, статус: {response.status}")
            return (name, len(content), response.status)
    except Exception as e:
        print(f"Ошибка при загрузке {name}: {e}")
        return (name, 0, f"Ошибка: {e}")

async def task4_async_scraper():
    """Основная асинхронная функция"""
    urls = [
        ("https://httpbin.org/delay/1", "Сайт 1"),
        ("https://httpbin.org/delay/2", "Сайт 2"),
        ("https://httpbin.org/delay/1", "Сайт 3"),
        ("https://httpbin.org/delay/3", "Сайт 4")
    ]

    start_time = time.time()

    async with aiohttp.ClientSession() as session:
        tasks = [fetch_url(session, url, name) for url, name in urls]
        results = await asyncio.gather(*tasks)

    end_time = time.time()

    total_time = end_time - start_time
    print(f"\n РЕЗУЛЬТАТЫ:")
    print(f"Общее время выполнения: {total_time:.2f} секунд")
    print(f"\nРазмеры контента:")
    for name, size, status in results:
        print(f"  {name}: {size} символов (статус: {status})")

    return results
    await task4_async_scraper()
```

```python
#Q5
import time
import threading
import multiprocessing
import asyncio

def io_task(name, duration):
    time.sleep(duration)
    return name

async def async_task(name, duration):
    await asyncio.sleep(duration)
    return name

async def run_async_part():
    """Запуск только асинхронной части"""
    tasks = [("Task1", 2), ("Task2", 3), ("Task3", 1), ("Task4", 2), ("Task5", 1)]
    coroutines = [async_task(name, duration) for name, duration in tasks]
    return await asyncio.gather(*coroutines)

def task5_fixed():
    tasks = [("Task1", 2), ("Task2", 3), ("Task3", 1), ("Task4", 2), ("Task5", 1)]

    print("ЗАДАЧА 5: СРАВНЕНИЕ ПРОИЗВОДИТЕЛЬНОСТИ")

    # 1. Синхронный
    print("\n1. Синхронный:")
    start = time.time()
    for n, s in tasks:
        io_task(n, s)
    t1 = time.time() - start
    print(f"   Время: {t1:.2f} сек")

    # 2. Многопоточный
    print("\n2. Многопоточный:")
    start = time.time()
    threads = [threading.Thread(target=io_task, args=(n, s)) for n, s in tasks]
    for t in threads: t.start()
    for t in threads: t.join()
    t2 = time.time() - start
    print(f"   Время: {t2:.2f} сек")

    # 3. Многопроцессный
    print("\n3. Многопроцессный:")
    start = time.time()
    processes = [multiprocessing.Process(target=io_task, args=(n, s)) for n, s in tasks]
    for p in processes: p.start()
    for p in processes: p.join()
    t3 = time.time() - start
    print(f"   Время: {t3:.2f} сек")

    # 4. Асинхронный - специально для Colab
    print("\n4. Асинхронный:")

    # Способ для Colab
    async def async_wrapper():
        start_async = time.time()
        await run_async_part()
        return time.time() - start_async

    # Запускаем в существующем event loop
    try:
        # Пробуем получить текущий loop
        loop = asyncio.get_event_loop()
        if loop.is_running():
            # В Colab loop уже запущен, запускаем задачу
            import nest_asyncio
            nest_asyncio.apply()
            t4 = loop.run_until_complete(async_wrapper())
        else:
            t4 = asyncio.run(async_wrapper())
    except:
        # Если не получилось, используем новый loop
        loop = asyncio.new_event_loop()
        asyncio.set_event_loop(loop)
        t4 = loop.run_until_complete(async_wrapper())

    print(f"   Время: {t4:.2f} сек")

    print(f"\n{'Метод':<12} {'Время':<10} {'Ускорение':<10}")
    print("-" * 35)
    print(f"{'Синхронный':<12} {t1:<10.2f} {'1.00x':<10}")
    print(f"{'Потоки':<12} {t2:<10.2f} {t1/t2:<10.2f}x")
    print(f"{'Процессы':<12} {t3:<10.2f} {t1/t3:<10.2f}x")
    print(f"{'Асинхронный':<12} {t4:<10.2f} {t1/t4:<10.2f}x")

    print(f"\nВывод: для I/O-bound задач лучше всего {min(t2, t3, t4):.2f} сек")

task5_fixed()
```
